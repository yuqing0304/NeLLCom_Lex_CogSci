```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy.stats import pearsonr
```

```python
frequencies = pd.read_csv('color_frequencies.csv')
drifts = pd.read_csv('drift_epoch30_summary_itemwise.csv')
```

## SL-RL-

```python
drift_data = drifts[drifts['condition'] == 'SL-RL-']
merged = pd.merge(drift_data, frequencies, on='color_name')
merged['log_frequency'] = np.log(merged['frequency'])
print(merged)
```

LOG

```python
seeds = merged['seed'].unique()
seeds = np.sort(seeds)
sns.set(style='whitegrid')
fig, axes = plt.subplots(4, 3, figsize=(5 * 3, 5 * 4), constrained_layout=True)
axes = axes.flatten()
significance = {'seed': [], 'r': [], 'p':[]}
for i, seed in enumerate(seeds):
    ax = axes[i]
    subset = merged[merged['seed'] == seed]
    r, p = pearsonr(subset['log_frequency'], subset['drift'])
    significance['seed'].append(seed)
    significance['r'].append(r)
    significance['p'].append(p)
    sns.scatterplot(
        data=subset,
        x='log_frequency',
        y='drift',
        hue='color_name',
        ax=ax,
        palette='tab20',
        legend=False
    )
    for _, row in subset.iterrows():
        ax.text(row['log_frequency'], row['drift'], row['color_name'], fontsize=8, alpha=0.7)

for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.suptitle('Color Drift vs log(Frequency) by Seed for SL-RL- Condition', fontsize=16)
plt.show()
print(pd.DataFrame(significance))
```
